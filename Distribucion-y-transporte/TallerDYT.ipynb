```python
#Nube de puntos de la imagen dada por VRP BOARD
import cv2
import numpy as np

img = cv2.imread("ruta.png")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Definir rangos HSV para cada color
lower_red = np.array([0, 100, 100])
upper_red = np.array([10, 255, 255])

lower_green = np.array([40, 50, 50])
upper_green = np.array([80, 255, 255])

lower_blue = np.array([100, 50, 50])
upper_blue = np.array([140, 255, 255])

# Máscaras
mask_red   = cv2.inRange(hsv, lower_red, upper_red)
mask_green = cv2.inRange(hsv, lower_green, upper_green)
mask_blue  = cv2.inRange(hsv, lower_blue, upper_blue)

# Función para obtener centros de puntos
def detectar_centros(mask, color_name):
    coords = []
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    for c in contours:
        ((x, y), radius) = cv2.minEnclosingCircle(c)
        if radius > 4:  # descartar ruido
            coords.append((int(x), int(y)))
    print(f"{color_name} encontrados:", len(coords))
    return coords

rojos  = detectar_centros(mask_red, "Rojos")
verdes = detectar_centros(mask_green, "Verdes")
azules = detectar_centros(mask_blue, "Azules")

# Mostrar todas las coordenadas
print("Coordenadas Rojos:", rojos)
print("Coordenadas Verdes:", verdes)
print("Coordenadas Azules:", azules)
```

```python
!pip install osmnx

import osmnx as ox
import networkx as nx
import folium

city = "Madrid, Spain"

G = ox.graph_from_place(city, network_type="drive")
G = ox.add_edge_speeds(G)
G = ox.add_edge_travel_times(G)

center_lat = 40.4168
center_lon = -3.7038
scale = 0.05

# Adding normalization steps here to ensure columns exist
df_clean["x_norm"] = (df_clean["x"] - df_clean["x"].min()) / (df_clean["x"].max() - df_clean["x"].min())
df_clean["y_norm"] = (df_clean["y"] - df_clean["y"].min()) / (df_clean["y"].max() - df_clean["y"].min())

df_clean["lat"] = center_lat + (df_clean["y_norm"] - 0.5)*scale
df_clean["lon"] = center_lon + (df_clean["x_norm"] - 0.5)*scale

m = folium.Map(location=(center_lat, center_lon), zoom_start=12)

colors = {"Rojo":"red","Verde":"green","Azul":"blue"}

for vehiculo in df_clean["color"].unique():

    subset = df_clean[df_clean["color"]==vehiculo]
    nodes = []

    for _,row in subset.iterrows():
        node = ox.distance.nearest_nodes(G, row["lon"], row["lat"])
        nodes.append(node)

    total_dist = 0
    total_time = 0
    full_route = []

    for i in range(len(nodes)-1):
        try:
            segment = nx.shortest_path(G, nodes[i], nodes[i+1], weight="length")
            full_route += segment

            total_dist += nx.shortest_path_length(G, nodes[i], nodes[i+1], weight="length")
            total_time += nx.shortest_path_length(G, nodes[i], nodes[i+1], weight="travel_time")
        except nx.NetworkXNoPath:
            print(f"No path found between nodes {nodes[i]} and {nodes[i+1]} for vehicle {vehiculo}. Skipping this segment.")

    if full_route:
        coords = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in full_route]
        folium.PolyLine(coords, color=colors[vehiculo], weight=4).add_to(m)

    print(f"\nVehículo {vehiculo}")
    print(f"Distancia real: {total_dist/1000:.2f} km")
    print(f"Tiempo estimado: {total_time/60:.2f} minutos")

m
```
