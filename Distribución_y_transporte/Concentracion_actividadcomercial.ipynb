```python
import folium
from folium.plugins import HeatMap

# 1. Configuración del mapa centrado en Cartagena
# Ajustado para abarcar desde la zona norte hasta Mamonal
m = folium.Map(location=[10.410, -75.500], zoom_start=12, tiles='OpenStreetMap')

# 2. Datos del Mapa de Calor (Latitud, Longitud, Intensidad)
heat_data = [
    # --- ZONA NORTE (Corregida) ---
    [10.5056, -75.4727, 0.7], # Serena del Mar
    [10.5153, -75.4700, 1.0], # C.C. Las Ramblas (Ubicado después de Serena del Mar)

    # --- PUERTO DE CARTAGENA (MANGA) ---
    [10.4053, -75.5287, 1.0], [10.4056, -75.5333, 0.9], # Sociedad Portuaria Manga

    # --- ZONA INDUSTRIAL (MAMONAL) - Densidad continua ---
    [10.365, -75.515, 0.8], [10.355, -75.508, 0.9], [10.345, -75.505, 1.0],
    [10.335, -75.503, 1.0], [10.325, -75.508, 1.0], [10.315, -75.515, 0.9],
    [10.305, -75.522, 0.9], [10.295, -75.530, 0.8], [10.285, -75.538, 0.7],

    # --- CENTROS COMERCIALES ---
    [10.4214, -75.5365, 1.0], # Mall Plaza El Castillo
    [10.4182, -75.5265, 1.0], # Caribe Plaza
    [10.3905, -75.4915, 1.0], # C.C. Paseo de la Castellana
    [10.3920, -75.4745, 1.0], # C.C. Multicentro La Plazuela
    [10.4020, -75.5565, 0.9], # Plaza Bocagrande


    # --- AEROPUERTO RAFAEL NÚÑEZ ---
    [10.448, -75.514, 0.8],

    # --- CENTRO HISTÓRICO Y TURISMO ---
    [10.423, -75.551, 1.0], # Centro Amurallado
    [10.405, -75.555, 0.9], # Bocagrande
]

# 3. Añadir el mapa de calor con parámetros de fusión
HeatMap(heat_data, radius=30, blur=20, max_zoom=10).add_to(m)

# 4. Etiquetas de ubicación (Marcadores con Tooltip)
puntos_referencia = [
    {"n": "C.C. Las Ramblas", "c": [10.5153, -75.4700], "i": "shopping-cart"},
    {"n": "Serena del Mar", "c": [10.5056, -75.4727], "i": "building"},
    {"n": "Puerto de Cartagena (Manga)", "c": [10.4053, -75.5287], "i": "anchor"},
    {"n": "Mall Plaza El Castillo", "c": [10.4214, -75.5365], "i": "shopping-cart"},
    {"n": "Caribe Plaza", "c": [10.4182, -75.5265], "i": "shopping-bag"},
    {"n": "C.C. La Castellana", "c": [10.3905, -75.4915], "i": "shopping-basket"},
    {"n": "C.C. La Plazuela", "c": [10.3920, -75.4745], "i": "shopping-cart"},
    {"n": "Corredor Industrial Mamonal", "c": [10.325, -75.508], "i": "industry"},
    {"n": "Aeropuerto", "c": [10.448, -75.514], "i": "plane"},
    {"n": "Centro Histórico", "c": [10.423, -75.551], "i": "fort-awesome"},
    {"n": "Bocagrande", "c": [10.405, -75.555], "i": "building"} # Agregado Bocagrande
]

for loc in puntos_referencia:
    folium.Marker(
        location=loc["c"],
        tooltip=loc["n"],
        icon=folium.Icon(color='blue', icon=loc["i"], prefix='fa')
    ).add_to(m)

# 5. Visualizar el mapa
m
```
